```python
import os
import glob
import zipfile

# Base directories
BASE_DIR = "/projects/structures/clusters/MHCI_align_07022025"
PDB_DIR = os.path.join(BASE_DIR, "all_pdb")
CONTACTS_DIR = os.path.join(BASE_DIR, "structure")
CA_DIR = os.path.join(BASE_DIR, "ca_coordinates_all")
ARCHIVE_DIR = os.path.join(BASE_DIR, "total_archives")

os.makedirs(ARCHIVE_DIR, exist_ok=True)

# Find all pdb files and extract hashes
for pdb_path in glob.glob(os.path.join(PDB_DIR, "aligned_aligned_*.pdb")):
    filename = os.path.basename(pdb_path)
    hash_id = filename.replace("aligned_aligned_", "").replace(".pdb", "")
    
    contacts_path = os.path.join(CONTACTS_DIR, f"{hash_id}_contacts.txt")
    ca_path = os.path.join(CA_DIR, f"{hash_id}_aa_coordinates.tsv")
    
    if os.path.exists(contacts_path) and os.path.exists(ca_path):
        archive_path = os.path.join(ARCHIVE_DIR, f"{hash_id}_all.zip")
        
        with zipfile.ZipFile(archive_path, 'w', zipfile.ZIP_DEFLATED) as z:
            z.write(pdb_path, os.path.basename(pdb_path))
            z.write(contacts_path, os.path.basename(contacts_path))
            z.write(ca_path, os.path.basename(ca_path))
```
